```python
%sql
-- Task 4: The ten accounts that matter most
SELECT
    c.trading_name,
    c.segment,
    c.province,
    COALESCE(r.rep_name, 'Unassigned')  AS sales_rep,
    ROUND(SUM(l.line_revenue), 2)       AS fy2025_revenue
FROM sizwe_customers c
JOIN sizwe_sales_orders o ON o.customer_id = c.customer_id
JOIN sizwe_sales_lines  l ON l.order_id    = o.order_id
LEFT JOIN sizwe_reps    r ON r.rep_id      = c.rep_id      -- LEFT JOIN so customers without a rep are kept
WHERE o.order_date > DATE '2024-06-30'
GROUP BY c.customer_id, c.trading_name, c.segment, c.province, r.rep_name
ORDER BY SUM(l.line_revenue) DESC
LIMIT 10;
-- Note: in this dataset every customer has a rep, but the LEFT JOIN + COALESCE protects against gaps.
```
