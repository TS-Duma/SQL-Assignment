```python
%sql
-- Task 3: The collections problem
SELECT
    c.channel,
    COUNT(*)                                                         AS invoices,
    COUNT(a.settlement_date)                                         AS settled,
   -- COUNT(*) - COUNT(a.settlement_date)                              AS still_open,
    ROUND(AVG(DATEDIFF(a.settlement_date, a.invoice_date)), 1)       AS avg_days_to_settle,
    ROUND(SUM(CASE WHEN a.settlement_date IS NULL
                   THEN a.invoice_amount - a.amount_settled END), 2) AS outstanding_value
FROM sizwe_ar_ledger a
JOIN sizwe_customers c ON c.customer_id = a.customer_id
GROUP BY c.channel
ORDER BY c.channel;
-- Answer: The problem sits in Wholesale: it takes 68.7 days on average to settle, against 3.8 for Subscription. Therefore; Wholesale is the channel to investigate first. 
```

```python
%sql
-- Task 3: The collections problem
-- Databricks note: DATEDIFF(end, start) returns whole days; plain date subtraction is not used here.
SELECT
    c.channel,
    COUNT(*)                                                         AS invoices,
    COUNT(a.settlement_date)                                         AS settled,
    COUNT(*) - COUNT(a.settlement_date)                              AS still_open,
    ROUND(AVG(DATEDIFF(a.settlement_date, a.invoice_date)), 1)       AS avg_days_to_settle,
    ROUND(SUM(CASE WHEN a.settlement_date IS NULL
                   THEN a.invoice_amount - a.amount_settled END), 2) AS outstanding_value
FROM sizwe_ar_ledger a
JOIN sizwe_customers c ON c.customer_id = a.customer_id
GROUP BY c.channel
ORDER BY c.channel;
-- Answer: The problem sits in Wholesale: it takes 68.7 days on average to settle (against 3.8 for Subscription),
-- and holds 3,454 of the 3,498 open invoices worth R87.8m of the R88.6m outstanding. Splitting by year
-- shows it is getting worse (Wholesale averaged 56.9 days in FY2024 and 84.1 in FY2025, on 60-day terms),
-- so Wholesale is the channel to investigate first.
```
