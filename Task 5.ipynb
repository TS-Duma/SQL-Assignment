```python
%sql
-- Task 5: A commercial gap
SELECT
    c.trading_name,
    c.segment,
    c.province,
    ROUND(SUM(l.line_revenue), 2) AS total_revenue
FROM sizwe_customers c
JOIN sizwe_sales_orders o ON o.customer_id = c.customer_id
JOIN sizwe_sales_lines  l ON l.order_id    = o.order_id
WHERE c.channel = 'Wholesale'
  AND NOT EXISTS (SELECT 1
                  FROM sizwe_trade_promotions t
                  WHERE t.customer_id = c.customer_id)
GROUP BY c.customer_id, c.trading_name, c.segment, c.province
ORDER BY SUM(l.line_revenue) DESC
LIMIT 15;
-- Answer: Have each account's sales rep meet Umhlanga Fresh 108 (R6.2m), Ridgeway Foods 308 (R4.4m) and Kalahari Kitchen 42 (R4.4m) and agree a first trade promotion, since all three are large buyers (all three are also in the FY2025 top ten) with no promotional support at all. (Revenue is across both years; 200 of the 600 Wholesale customers have never had a promotion.)
```
