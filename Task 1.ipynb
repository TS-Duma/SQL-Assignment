```python
%sql 
-- Task 1: Channel performance

SELECT
    o.channel,
    ROUND(SUM(CASE WHEN o.order_date <= DATE '2024-06-30' THEN l.line_revenue END), 2) AS fy2024_revenue,
    ROUND(SUM(CASE WHEN o.order_date >  DATE '2024-06-30' THEN l.line_revenue END), 2) AS fy2025_revenue,
    ROUND(100.0 * (
          SUM(CASE WHEN o.order_date >  DATE '2024-06-30' THEN l.line_revenue END)
        - SUM(CASE WHEN o.order_date <= DATE '2024-06-30' THEN l.line_revenue END))
        / SUM(CASE WHEN o.order_date <= DATE '2024-06-30' THEN l.line_revenue END), 1) AS growth_pct
FROM sizwe_sales_orders o
JOIN sizwe_sales_lines  l ON l.order_id = o.order_id
GROUP BY o.channel
ORDER BY o.channel;
-- Answer: Wholesale grew from R308.2m to R335.4m and Subscription from R132.8m to R144.6m, both +8.8%. Because both channels grew at the same rate, the split has not moved (Wholesale 69.9% / Subscription 30.1% in both years), so growth is not coming from a channel shift.
```
