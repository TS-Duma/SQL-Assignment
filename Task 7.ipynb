```python
%sql
-- Task 7: The shape of the year
WITH monthly AS (
    SELECT
        DATE_TRUNC('month', o.order_date)::date AS month,
        SUM(l.line_revenue)                     AS revenue
    FROM sizwe_sales_orders o
    JOIN sizwe_sales_lines  l ON l.order_id = o.order_id
    GROUP BY 1
)
SELECT
    month,
    ROUND(revenue, 2) AS revenue,
    ROUND(100.0 * (revenue - LAG(revenue) OVER (ORDER BY month))
          / LAG(revenue) OVER (ORDER BY month), 1)                  AS mom_change_pct,
    ROUND(AVG(revenue) OVER (ORDER BY month
          ROWS BETWEEN 2 PRECEDING AND CURRENT ROW), 2)             AS rolling_3m_avg
FROM monthly
ORDER BY month;
-- Answer: Yes, the pattern repeats in both years: revenue peaks in June/July (R41.5m-R45.1m), drops for the second half of the year and then recovers from February. December is the quietest month in both years (R29.9m in Dec 2023 and R32.6m in Dec 2024, a fall of about 13% on November each time).
```
