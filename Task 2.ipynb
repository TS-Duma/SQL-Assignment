```python
%sql
-- Task 2: Where the profit actually comes from
SELECT
    p.product_line,
    ROUND(SUM(l.line_revenue), 2)                              AS revenue,
    ROUND(SUM(l.cogs), 2)                                      AS cost_of_goods,
    ROUND(SUM(l.line_revenue - l.cogs), 2)                     AS gross_profit,
    ROUND(100.0 * SUM(l.line_revenue - l.cogs)
          / SUM(l.line_revenue), 1)                            AS gross_margin_pct
FROM sizwe_sales_lines l
JOIN sizwe_products    p ON p.sku = l.sku
GROUP BY p.product_line
ORDER BY gross_profit DESC;
-- Answer: House Blend earns the most profit (R107.1m at 38.3%) but it is not the highest-margin line. Single Origin has the best margin (44.3%) on a smaller base (R75.3m profit). Bulk Commercial is the second-biggest revenue line (R271.8m) but only makes a 19.9% margin, so it earns far less profit per rand.
```
