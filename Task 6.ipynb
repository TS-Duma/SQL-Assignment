```python
%sql
-- Task 6: Are the sales representatives hitting target?
SELECT
    r.rep_name,
    r.territory,
    r.target_monthly,
    ROUND(COALESCE(x.fy25_revenue, 0) / 12, 2)                            AS fy2025_monthly_avg,
    ROUND(100.0 * COALESCE(x.fy25_revenue, 0) / 12 / r.target_monthly, 1) AS pct_of_target
FROM sizwe_reps r
LEFT JOIN (
    SELECT o.rep_id, SUM(l.line_revenue) AS fy25_revenue
    FROM sizwe_sales_orders o
    JOIN sizwe_sales_lines  l ON l.order_id = o.order_id
    WHERE o.order_date > DATE '2024-06-30'
    GROUP BY o.rep_id
) x ON x.rep_id = r.rep_id
ORDER BY pct_of_target DESC;
-- Answer: 16 of 18 reps beat their target; only Sipho Steyn (98.0%) and Johan Naidoo (77.6%) fall short. Several reps are at 200%+ (Karabo van Wyk 250.3%, Kagiso Mkhize 249.8%), which suggests the targets are set too low for some territories. Kagiso's figure is helped by Halvane Retail Group (R38.4m in FY2025).
```
